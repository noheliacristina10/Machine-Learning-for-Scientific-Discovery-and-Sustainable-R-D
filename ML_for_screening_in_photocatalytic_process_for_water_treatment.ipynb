```python
#@title Analysis functions — part 1
from pathlib import Path
from urllib.request import urlretrieve
import hashlib, json, warnings
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, Markdown, HTML
from sklearn.base import clone
from sklearn.compose import ColumnTransformer, TransformedTargetRegressor
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.model_selection import GroupKFold
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error, median_absolute_error
from sklearn.dummy import DummyRegressor
from sklearn.linear_model import Ridge, ElasticNet
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor, ExtraTreesRegressor, GradientBoostingRegressor, BaggingRegressor
from sklearn.neural_network import MLPRegressor
from sklearn.exceptions import ConvergenceWarning
from catboost import CatBoostRegressor
from scipy.stats import qmc
from scipy.spatial.distance import cdist
from pymoo.util.nds.non_dominated_sorting import NonDominatedSorting

# Cambria-style document typography for charts and tabular outputs.
from matplotlib import font_manager
CAMBRIA_FONT_URL = "https://raw.githubusercontent.com/google/fonts/main/ofl/caladea/Caladea-Regular.ttf"
CAMBRIA_FONT_PATH = Path("Caladea-Regular.ttf")
try:
    if not CAMBRIA_FONT_PATH.exists():
        urlretrieve(CAMBRIA_FONT_URL, CAMBRIA_FONT_PATH)
    font_manager.fontManager.addfont(str(CAMBRIA_FONT_PATH))
    CHART_FONT = font_manager.FontProperties(fname=str(CAMBRIA_FONT_PATH)).get_name()
except Exception:
    CHART_FONT = "DejaVu Serif"

CAMBRIA_STACK = "Cambria, Caladea, Georgia, Times New Roman, serif"
display(HTML(f"""
<style>
@import url('https://fonts.googleapis.com/css2?family=Caladea:ital,wght@0,400;0,700;1,400&display=swap');
.output_html, .output_text, table, th, td {{ font-family: {CAMBRIA_STACK} !important; }}
</style>
"""))
plt.rcParams.update({
    "font.family": CHART_FONT,
    "mathtext.fontset": "stix",
    "axes.titleweight": "semibold",
    "figure.titleweight": "semibold",
    "font.size": 11,
})

REFERENCES={'Diazinon':'10.1016/j.seppur.2007.07.016','Dichlorvos':'10.1016/j.apcatb.2005.01.005','Phenol':'10.1016/j.aej.2013.04.005'}
SCENARIO=['Pollutant','Type of dopant','Light source']
DECISIONS=['pH','Catalyst dosage (mg/L)','t (h)','Initial C of pollutant (mg/L)','Weight ratio of doping element to Zn']
FEATURES=['Light source','Weight ratio of doping element to Zn','Mw dop','Mw Poll','Sol dop','Sol Poll',
          'Initial C of pollutant (mg/L)','pH','Catalyst dosage (mg/L)','t (h)','Pollutant','Type of dopant']
NUMERIC=[c for c in FEATURES if c not in SCENARIO]
TARGET='Degradation (%)'; TIME='t (h)'; DOSE='Catalyst dosage (mg/L)'; INLET='Initial C of pollutant (mg/L)'
SERIES_KEYS=SCENARIO+['Weight ratio of doping element to Zn','Mw dop','Mw Poll','Sol dop','Sol Poll',INLET,'pH',DOSE]
DATA_URL=('https://raw.githubusercontent.com/noheliacristina10/ML-for-chemical-sustainability/'
          '8d8aeca37f2be53e4efeacbb79cfa49bc1d863bc/data/raw/'
          'Photocatalytic%20Degradation%20of%20Organic%20Pollutants.xlsx')
EXPECTED_SHA='778a0ef60289d4d0887d5959cd38368a3efa21dcd8b8d5ba06749d36ca658423'

plt.rcParams.update({'figure.dpi':120,'axes.spines.top':False,'axes.spines.right':False,'font.size':10})

def compact_table(frame,decimals=3):
    display(frame.style.format(precision=decimals,na_rep='—').hide(axis='index')
      .set_properties(**{'text-align':'left','padding':'6px 10px'})
      .set_table_styles([{'selector':'th','props':[('text-align','left'),('border-bottom','2px solid #166b72'),('padding','6px 10px')]}]))

def load_data(path):
    path=Path(path)
    if not path.exists():urlretrieve(DATA_URL,path)
    digest=hashlib.sha256(path.read_bytes()).hexdigest()
    if digest!=EXPECTED_SHA:raise ValueError(f'Workbook SHA-256 mismatch: {digest}')
    data=pd.read_excel(path);data.columns=data.columns.str.strip()
    for c in SCENARIO:data[c]=data[c].astype(str).str.strip()
    data['Type of dopant']=data['Type of dopant'].replace('-','None')
    data['Light source']=data['Light source'].replace({'1':'UV','2':'Visible'})
    refs=data.Reference.astype('string').str.strip().str.lower().str.replace(r'^https?://(dx\.)?doi\.org/','',regex=True)
    if refs.isna().any() or refs.isin(['','nan','none','unknown','-']).any():raise ValueError('Resolve missing references.')
    data['Reference']=refs
    for c in NUMERIC+[TARGET]:data[c]=pd.to_numeric(data[c],errors='raise')
    if not np.isfinite(data[NUMERIC+[TARGET]].to_numpy()).all():raise ValueError('Missing/nonfinite values.')
    if not data[TARGET].between(0,100).all():raise ValueError('Observed response outside 0–100%.')
    data['Source row']=np.arange(2,len(data)+2)
    data['Scenario']=data[SCENARIO].agg(' | '.join,axis=1)
    return data.reset_index(drop=True),digest

def metrics(y,pred):
    y=np.asarray(y,float);pred=np.asarray(pred,float)
    return {'R²':r2_score(y,pred) if len(y)>1 and np.ptp(y)>0 else np.nan,
            'RMSE':np.sqrt(mean_squared_error(y,pred)),'MAE':mean_absolute_error(y,pred),
            'Median AE':median_absolute_error(y,pred),
            'Outside 0–100 (%)':100*np.mean((pred<0)|(pred>100))}

```

```python
def preprocessor():
    return ColumnTransformer([('numeric',StandardScaler(),NUMERIC),
        ('categorical',OneHotEncoder(handle_unknown='ignore',sparse_output=False),SCENARIO)])

def learned_models(seed=42):
    base=DecisionTreeRegressor(max_depth=3,min_samples_leaf=5,random_state=seed)
    specs={
      'Extra Trees':ExtraTreesRegressor(n_estimators=300,max_depth=6,min_samples_leaf=5,n_jobs=2,random_state=seed),
      'Bagged shallow trees':BaggingRegressor(estimator=base,n_estimators=100,max_samples=.8,n_jobs=1,random_state=seed),
      'Shallow Gradient Boosting':GradientBoostingRegressor(n_estimators=300,max_depth=2,min_samples_leaf=5,learning_rate=.03,random_state=seed),
      'Random Forest':RandomForestRegressor(n_estimators=300,max_depth=6,min_samples_leaf=5,n_jobs=2,random_state=seed),
      'Elastic Net':ElasticNet(alpha=1.,l1_ratio=.2,max_iter=20000,tol=1e-5),
      'CatBoost':CatBoostRegressor(iterations=500,depth=6,learning_rate=.03,l2_leaf_reg=5,loss_function='RMSE',
          random_seed=seed,verbose=False,allow_writing_files=False,thread_count=2),
      'Shallow decision tree':DecisionTreeRegressor(max_depth=3,min_samples_leaf=5,random_state=seed),
      'Small ANN':TransformedTargetRegressor(regressor=MLPRegressor(hidden_layer_sizes=(16,),activation='tanh',
          solver='lbfgs',alpha=1.,max_iter=5000,max_fun=50000,tol=1e-6,random_state=seed),transformer=StandardScaler()),
      'Ridge':Ridge(alpha=10.)}
    return {name:Pipeline([('preprocess',preprocessor()),('model',model)]) for name,model in specs.items()}

def series_splits(dataset):
    series=dataset.groupby(SERIES_KEYS,dropna=False,sort=True).ngroup()
    n_splits=min(5,series.nunique())
    if n_splits<3:raise ValueError('At least three proxy kinetic series are required.')
    splits=list(GroupKFold(n_splits).split(dataset,groups=series))
    for train,test in splits:assert set(series.iloc[train]).isdisjoint(series.iloc[test])
    return series,splits

def compare_models(dataset,pesticide):
    series,splits=series_splits(dataset);pred_parts=[];score_rows=[];warning_rows=[];fitted={}
    for fold,(train,test) in enumerate(splits,1):
        training,assessment=dataset.iloc[train],dataset.iloc[test]
        candidates={'Training mean':DummyRegressor(strategy='mean'),
                    'Training median':DummyRegressor(strategy='median'),**learned_models(42+fold)}
        for name,model in candidates.items():
            if name.startswith('Training '):
                model.fit(np.zeros((len(training),1)),training[TARGET]);pred=model.predict(np.zeros((len(assessment),1)))
            else:
                with warnings.catch_warnings(record=True) as caught:
                    warnings.simplefilter('always');model.fit(training[FEATURES],training[TARGET])
                warning_rows.extend({'Pesticide':pesticide,'Fold':fold,'Model':name,
                    'Category':w.category.__name__,'Message':str(w.message)} for w in caught)
                scaler=model.named_steps['preprocess'].named_transformers_['numeric']
                assert np.allclose(scaler.mean_,training[NUMERIC].mean())
                pred=model.predict(assessment[FEATURES]);fitted[(fold,name)]=model
            rows=assessment[['Source row','Reference','Scenario']].copy()
            rows['Observed']=assessment[TARGET].to_numpy();rows['Predicted']=pred
            rows['Pesticide']=pesticide;rows['Fold']=fold;rows['Model']=name;rows['Series']=series.iloc[test].to_numpy()
            pred_parts.append(rows);score_rows.append({'Pesticide':pesticide,'Fold':fold,'Model':name,'Rows':len(test),**metrics(rows.Observed,pred)})
        print(f'{pesticide}: fold {fold}/{len(splits)} complete',flush=True)
    predictions=pd.concat(pred_parts,ignore_index=True);scores=pd.DataFrame(score_rows)
    summary=scores.groupby(['Pesticide','Model']).agg(**{'Mean fold MAE':('MAE','mean'),'SD fold MAE':('MAE','std'),
        'Mean fold R²':('R²','mean'),'Mean fold RMSE':('RMSE','mean')}).reset_index()
    pooled=[]
    for (pest,name),g in predictions.groupby(['Pesticide','Model']):pooled.append({'Pesticide':pest,'Model':name,**metrics(g.Observed,g.Predicted)})
    summary=summary.merge(pd.DataFrame(pooled),on=['Pesticide','Model']).sort_values(['Mean fold MAE','MAE'],kind='stable').reset_index(drop=True)
    return series,splits,predictions,scores,summary,pd.DataFrame(warning_rows),fitted

def fit_family(dataset,model_name,seed=42):
    model=learned_models(seed)[model_name]
    with warnings.catch_warnings(record=True) as caught:
        warnings.simplefilter('always');model.fit(dataset[FEATURES],dataset[TARGET])
    return model,caught
```

```python
#@title Analysis functions — part 3
def support_geometry(observed):
    active=[c for c in DECISIONS if observed[c].nunique()>1]
    mins=observed[active].min().to_numpy(float);ranges=np.ptp(observed[active].to_numpy(float),axis=0)
    scaled=(observed[active].to_numpy(float)-mins)/ranges
    distances=cdist(scaled,scaled);np.fill_diagonal(distances,np.inf)
    return active,mins,ranges,scaled,distances.min(axis=1)

def predict_candidates(rows,final_model,ensemble):
    final=np.asarray(final_model.predict(rows[FEATURES]),float)
    members=np.column_stack([m.predict(rows[FEATURES]) for m in ensemble])
    disagreement=members.std(axis=1,ddof=1)
    return final,disagreement,final-1.645*disagreement

def support_distance(rows,observed):
    active,mins,ranges,scaled,loo=support_geometry(observed)
    candidate=(rows[active].to_numpy(float)-mins)/ranges
    return cdist(candidate,scaled).min(axis=1),loo,active

def nondominated(frame):
    objectives=np.column_stack([-frame['Conservative degradation (%)'],frame[TIME],frame[DOSE]])
    indices=NonDominatedSorting().do(objectives,only_non_dominated_front=True)
    return frame.iloc[indices].copy().reset_index(drop=True)

def choose_knee(front):
    objectives=np.column_stack([-front['Conservative degradation (%)'],front[TIME],front[DOSE]])
    ranges=np.ptp(objectives,axis=0);ranges[ranges==0]=1
    normalized=(objectives-objectives.min(axis=0))/ranges
    distance=np.sqrt((normalized**2).mean(axis=1))
    chosen=front.iloc[int(np.argmin(distance))].copy();chosen['Knee distance']=distance.min()
    return chosen

def scenario_candidates(dataset,scenario,n_power=15):
    observed=dataset[dataset.Scenario.eq(scenario)].reset_index(drop=True)
    template={c:observed.iloc[0][c] for c in FEATURES}
    variable=[c for c in DECISIONS if observed[c].nunique()>1]
    lower=observed[variable].min().to_numpy(float);upper=observed[variable].max().to_numpy(float)
    if len(variable)==1:
        values=np.linspace(lower[0],upper[0],4096)[:,None]
    else:
        values=qmc.scale(qmc.Sobol(len(variable),scramble=True,seed=2028).random_base2(n_power),lower,upper)
    rows=pd.DataFrame([template]*len(values))[FEATURES]
    for j,c in enumerate(variable):rows[c]=values[:,j]
    rows=pd.concat([rows,observed[FEATURES]],ignore_index=True).drop_duplicates().reset_index(drop=True)
    return rows,observed,variable,lower,upper

def screen_supported(dataset,scenario,final_model,ensemble,pesticide):
    rows,observed,variable,lower,upper=scenario_candidates(dataset,scenario)
    mean,disagreement,conservative=predict_candidates(rows,final_model,ensemble)
    rows['Predicted degradation (%)']=mean;rows['Model disagreement (%)']=disagreement
    rows['Conservative degradation (%)']=conservative
    rows['Nearest observed distance'],loo,active=support_distance(rows,observed)
    sensitivity=[];fronts={};knees={}
    for quantile in [.50,.75,.90]:
        threshold=float(np.quantile(loo,quantile));supported=rows[rows['Nearest observed distance']<=threshold]
        physical=supported[supported['Predicted degradation (%)'].between(0,100)]
        if physical.empty:
            sensitivity.append({'Pesticide':pesticide,'Quantile':quantile,'Threshold':threshold,'Candidate?':False});continue
        front=nondominated(physical);knee=choose_knee(front);fronts[quantile]=front;knees[quantile]=knee
        sensitivity.append({'Pesticide':pesticide,'Quantile':quantile,'Threshold':threshold,'Candidate?':True,
            'Candidates evaluated':len(rows),'Supported candidates':len(supported),'Physical candidates':len(physical),
            'Pareto candidates':len(front),'Unbounded outside 0–100 (%)':100*np.mean(~rows['Predicted degradation (%)'].between(0,100)),**{c:knee[c] for c in variable},
            'Predicted degradation (%)':knee['Predicted degradation (%)'],'Model disagreement (%)':knee['Model disagreement (%)'],
            'Conservative degradation (%)':knee['Conservative degradation (%)'],'Nearest observed distance':knee['Nearest observed distance']})
    if .75 not in knees:return rows,pd.DataFrame(),None,pd.DataFrame(sensitivity),variable,lower,upper
    chosen=knees[.75].copy();chosen['Pesticide']=pesticide;chosen['Scenario']=scenario
    chosen['Calculated residual (mg/L)']=chosen[INLET]*(1-chosen['Predicted degradation (%)']/100)
    chosen['Calculated removal rate (mg/L/h)']=chosen[INLET]*chosen['Predicted degradation (%)']/100/chosen[TIME]
    chosen['Calculated catalyst productivity (mg/mg)']=chosen[INLET]*chosen['Predicted degradation (%)']/100/chosen[DOSE]
    return rows,fronts[.75],chosen,pd.DataFrame(sensitivity),variable,lower,upper

```

```python
#@title Transparent sustainability proxy calculations
def add_sustainability_proxies(frame):
    frame=frame.copy()
    removed=frame[INLET]*frame['Predicted degradation (%)']/100
    frame['Pollutant removed (mg/L)']=removed
    frame['Residual pollutant (mg/L)']=frame[INLET]-removed
    frame['Removal rate (mg/L/h)']=removed/frame[TIME].replace(0,np.nan)
    frame['Catalyst productivity (mg/mg)']=removed/frame[DOSE].replace(0,np.nan)
    frame['Catalyst-time exposure (mg h/L)']=frame[DOSE]*frame[TIME]
    return frame
```
